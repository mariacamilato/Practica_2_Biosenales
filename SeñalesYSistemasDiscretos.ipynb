```python
import matplotlib
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 
```

```python
n = np.array([-2, -1, 0, 1, 2 ,3])
x = np.array([5, 6, 7, 5, 4, 3])
```

```python
#SEÑAL IMPULSO UNITARIO
def impseq(n0, n1, n2):
#   Genera x(n) = delta(n-n0); n1 <= n <= n2
#	 ----------------------------------------------
    n = np.arange(n1,n2+1) # Se crea el vector de muestras
    x = (n-n0) == 0
    return [x,n]
```

```python
x, n =impseq(0, -5, 5)
plt.stem(n,x)
plt.title("Impulso unitario")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid()
plt.show()
```

```python
#SEÑAL ESCALÓN UNITARO
def escauni(n0, n1, n2):
#   Genera x(n) = delta(n-n0); n1 <= n <= n2
#	 ----------------------------------------------
    n = np.arange(n1,n2+1) # Se crea el vector de muestras
    x = (n >= n0).astype(int)
    return [x,n]
```

```python
x, n =escauni(0, -5, 5)
plt.stem(n,x)
plt.title("Escalón unitario")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid()
plt.show()
```

```python
#SEÑAL RAMPA
def rampa(n0, n1, n2, m=1):
#   Genera x(n) = delta(n-n0); n1 <= n <= n2
#	 ----------------------------------------------
    n = np.arange(n1,n2+1) # Se crea el vector de muestras
    x = m* (n-n0) * (n >=n0)
    return [x,n]
```

```python
x, n =rampa(0, -5, 5, m=1)
plt.stem(n,x)
plt.title("Rampa")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid()
plt.show()
```

```python
#SEÑAL EXPONENCIAL COMPLEJA
def exponencial(s, c, n1, n2):
#   Genera x(n) = delta(n-n0); n1 <= n <= n2
#	 ----------------------------------------------
    n = np.arange(n1,n2+1) # Se crea el vector de muestras
    x = np.exp((s + 1j*c)*n)
    return [x,n]
```

```python
x, n =exponencial(-0.2, 0.5, 0, 20)
plt.stem(n, np.real(x))
plt.title("Exponencial compleja")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid()
plt.show()
```

```python
#SEÑAL SENOSOIDAL
def seno(A,k, phi, n1, n2):
    n=np.arange(n1, n2+1)
    x= A* np.sin(k*n + phi)
    return x,n
```

```python
x, n =seno(1, 0.3*np.pi, 0, 0 ,20)
plt.stem(n, x)
plt.title("Senosoidal")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid()
plt.show()
```

```python
 # 3.a SECUENCIA CON IMPULSOS

x1_1, n = impseq(-1, -4, 4) 
x1_2, _ = impseq(-3, -4, 4) 
x1_3, _ = impseq(-2, -4, 4)  
x1_4, _ = impseq(0,  -4, 4)  

x1 = 3*x1_1 + 5*x1_2 + 3*x1_3 + 3*x1_4 + x1_4  

plt.figure(figsize=(8, 4))
plt.stem(n, x1)
plt.title("Primera secuencia de impulsos")
plt.xlabel("n")
plt.ylabel("x_1[n]")
plt.xticks(n)
plt.grid(True)
plt.tight_layout()
plt.savefig("fig_3a.png", dpi=100)
plt.show()
```

```python
# 3.b SISTEMA CON ESCALONES Y ADELANTOS

u1, n2 = escauni(-4, -6, 6)   
u2, _  = escauni(-2, -6, 6)   
x2 = -2*u1 + 3*u2

plt.figure(figsize=(8, 4))
plt.stem(n2, x2)
plt.title("señal escalon adelantada")
plt.xlabel("n")
plt.ylabel("x_2[n]")
plt.xticks(n2)
plt.grid(True)
plt.tight_layout()
plt.savefig("fig_3b.png", dpi=100)
plt.show()
```

```python
# 3.c SISTEMA CON EXPONENCIAL Y SINUSOIDE (0 <= n <= 100)
n3 = np.arange(0, 101)
x3 = np.exp(0.1 * n3) * np.cos(0.2 * np.pi * n3)

plt.figure(figsize=(10, 4))
plt.stem(n3, x3, markerfmt='C0o', linefmt='C0-', basefmt='k-')
plt.title("exponencial y sinusoide")
plt.xlabel("n")
plt.ylabel("x_3[n]")
plt.grid(True)
plt.tight_layout()
plt.savefig("fig_3c.png", dpi=100)
plt.show()
```

```python
# 3.d COMBINACIÓN DE RAMPA Y ESCALÓN

r1, n4 = rampa(-3, -10, 10)  
r2, _  = rampa( 2, -10, 10)  
u3, _  = rampa( 3, -10, 10)  
x4 = 2*r1 - r2 - 5*u3

plt.figure(figsize=(10, 4))
plt.stem(n4, x4)
plt.title("Rampa y escalón")
plt.xlabel("n")
plt.ylabel("x_4[n]")
plt.xticks(n4[::2])
plt.grid(True)
plt.tight_layout()
plt.savefig("fig_3d.png", dpi=100)
plt.show()
```

```python

# 4)SECUENCIA POR TRAMOS 
n_seg1 = np.arange(0, 6)
n_seg2 = np.arange(6, 12)
n_seg3 = np.arange(12, 18)

x_seg1 = n_seg1                
x_seg2 = n_seg2 - 5           
x_seg3 = n_seg3 - 10           

x_tramos = np.concatenate((x_seg1, x_seg2, x_seg3))
n_tramos  = np.concatenate((n_seg1, n_seg2, n_seg3))

# 4.a)
dx_tramos  = np.diff(x_tramos)
n_dx_tramos = n_tramos[:-1]

# 4.b) 
fig, axes = plt.subplots(2, 1, figsize=(10, 7))

axes[0].stem(n_tramos, x_tramos)
axes[0].set_title("4) Secuencia por tramos x[n]")
axes[0].set_xlabel("n")
axes[0].set_ylabel("x[n]")
axes[0].grid(True)

axes[1].stem(n_dx_tramos, dx_tramos, linefmt='C1-', markerfmt='C1o', basefmt='k-')
axes[1].set_title("4.a)")
axes[1].set_xlabel("n")
axes[1].set_ylabel("\\Delta x[n]")
axes[1].grid(True)

plt.tight_layout()
plt.savefig("fig_4.png", dpi=100)
plt.show()
```
