```python
import matplotlib
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 
```

```python
n = np.array([-2, -1, 0, 1, 2 ,3])
x = np.array([5, 6, 7, 5, 4, 3])
```

```python
#SEÑAL IMPULSO UNITARIO
def impseq(n0, n1, n2):
#   Genera x(n) = delta(n-n0); n1 <= n <= n2
#	 ----------------------------------------------
    n = np.arange(n1,n2+1) # Se crea el vector de muestras
    x = (n-n0) == 0
    return [x,n]
```

```python
x, n =impseq(0, -5, 5)
plt.stem(n,x)
plt.title("Impulso unitario")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid()
plt.show()
```

```python
#SEÑAL ESCALÓN UNITARO
def escauni(n0, n1, n2):
#   Genera x(n) = delta(n-n0); n1 <= n <= n2
#	 ----------------------------------------------
    n = np.arange(n1,n2+1) # Se crea el vector de muestras
    x = (n >= n0).astype(int)
    return [x,n]
```

```python
x, n =escauni(0, -5, 5)
plt.stem(n,x)
plt.title("Escalón unitario")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid()
plt.show()
```

```python
#SEÑAL RAMPA
def rampa(n0, n1, n2, m=1):
#   Genera x(n) = delta(n-n0); n1 <= n <= n2
#	 ----------------------------------------------
    n = np.arange(n1,n2+1) # Se crea el vector de muestras
    x = m* (n-n0) * (n >=n0)
    return [x,n]
```

```python
x, n =rampa(0, -5, 5, m=1)
plt.stem(n,x)
plt.title("Rampa")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid()
plt.show()
```

```python
#SEÑAL EXPONENCIAL COMPLEJA
def exponencial(s, c, n1, n2):
#   Genera x(n) = delta(n-n0); n1 <= n <= n2
#	 ----------------------------------------------
    n = np.arange(n1,n2+1) # Se crea el vector de muestras
    x = np.exp((s + 1j*c)*n)
    return [x,n]
```

```python
x, n =exponencial(-0.2, 0.5, 0, 20)
plt.stem(n, np.real(x))
plt.title("Exponencial compleja")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid()
plt.show()
```

```python
#SEÑAL SENOSOIDAL
def seno(A,k, phi, n1, n2):
    n=np.arange(n1, n2+1)
    x= A* np.sin(k*n + phi)
    return x,n
```

```python
x, n =seno(1, 0.3*np.pi, 0, 0 ,20)
plt.stem(n, x)
plt.title("Senosoidal")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid()
plt.show()
```

```python
 #SECUENCIA CON IMPULSOS

x1_1, n = impseq(-1, -4, 4) 
x1_2, _ = impseq(-3, -4, 4) 
x1_3, _ = impseq(-2, -4, 4)  
x1_4, _ = impseq(0,  -4, 4)  

x1 = 3*x1_1 + 5*x1_2 + 3*x1_3 + 3*x1_4 + x1_4  

plt.figure(figsize=(8, 4))
plt.stem(n, x1)
plt.title("3.a) $x_1(n) = 3\\delta(n+1) + 5\\delta(n+3) + 3\\delta(n+2) + 4\\delta(n)$")
plt.xlabel("n")
plt.ylabel("$x_1[n]$")
plt.xticks(n)
plt.grid(True)
plt.tight_layout()
plt.savefig("fig_3a.png", dpi=100)
plt.show()
```

```python
#SISTEMA CON ESCALONES Y ADELANTOS

u1, n2 = escauni(-4, -6, 6)   
u2, _  = escauni(-2, -6, 6)   
x2 = -2*u1 + 3*u2

plt.figure(figsize=(8, 4))
plt.stem(n2, x2)
plt.title("3.b) $x_2(n) = -2u(n+4) + 3u(n+2)$")
plt.xlabel("n")
plt.ylabel("$x_2[n]$")
plt.xticks(n2)
plt.grid(True)
plt.tight_layout()
plt.savefig("fig_3b.png", dpi=100)
plt.show()
```
